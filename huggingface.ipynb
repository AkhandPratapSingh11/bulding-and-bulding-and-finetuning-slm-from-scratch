```python
import os
from dotenv import load_dotenv, find_dotenv

# Load .env file (checking current working directory and SLM_Experiment)
load_dotenv(find_dotenv(usecwd=True))
load_dotenv(".env")
load_dotenv("SLM_Experiment/.env")

READ_TOKEN = os.environ.get("HUGGINGFACE_READ_TOKEN")
WRITE_TOKEN = os.environ.get("HUGGINGFACE_WRITE_TOKEN")
FULL_ACCESS_TOKEN = os.environ.get("HUGGINGFACE_FULL_ACCESS_TOKEN")

print("Tokens loaded:")
print(f"- READ_TOKEN: {bool(READ_TOKEN)}")
print(f"- WRITE_TOKEN: {bool(WRITE_TOKEN)}")
print(f"- FULL_ACCESS_TOKEN: {bool(FULL_ACCESS_TOKEN)}")
```

```python
from huggingface_hub import login, HfApi

# Select token (prioritizing full access / write token)
token_to_use = FULL_ACCESS_TOKEN or WRITE_TOKEN or READ_TOKEN

if token_to_use:
    login(token=token_to_use, add_to_git_credential=True)
    print("Successfully logged in to Hugging Face!")
else:
    print("Error: No Hugging Face token found in environment variables.")
```

```python
api = HfApi()
user_info = api.whoami()
print(f"Logged in user: {user_info.get('name')}")
print(f"Account type: {user_info.get('type')}")
print(f"Email: {user_info.get('email')}")
```
